# Variables & Basic Data Types 📦
**Concept:**

Think of a variable as a labeled storage box. In Machine Learning, we constantly handle different kinds of data—like housing prices, house sizes, or whether a house has a garden. Python automatically recognizes the type of data stored in the box:


*   🔢 Integer (int): Whole numbers (e.g., number of rooms = 3)

*   ⚖️ Float (float): Decimal numbers (e.g., house price in millions = 1.5)

*   🔤 String (str): Text wrapped in quotes (e.g., city = "New York")

*   🔘 Boolean (bool): True or False values (e.g., has a garage = True)








```python
# --- 1. Creating Variables ---
house_rooms = 3             # int: whole number
house_price = 1.5           # float: decimal number
city_name = "New York"      # str: text
has_garage = True           # bool: True or False

# --- 2. Checking Types ---
# The type() function tells us what kind of data is stored in the variable
print("Data type of house_rooms:", type(house_rooms))
print("Data type of house_price:", type(house_price))
print("Data type of city_name:", type(city_name))
print("Data type of has_garage:", type(has_garage))

# --- 3. Simple Calculation ---
# Updating a price value
new_price = house_price + 0.2
print("Updated price:", new_price)
```

# 2. Lists & Indexing 📋
**Concept:**

A List is an ordered collection of items, written inside square brackets [...]. In ML, we often collect a series of measurements (like multiple house prices) before converting them into NumPy arrays.



*   0-based indexing: Python counts starting from index 0. The first item is at index 0, the second at 1, and so on.

*  Negative indexing: Index -1 gives the last item in the list.



```python
# --- 1. Creating a List of Feature Values ---
house_prices = [1.2, 1.5, 2.1, 0.9, 3.4]
print("All prices:", house_prices)

# --- 2. Accessing Items (Indexing) ---
first_price = house_prices[0]    # First item (index 0)
last_price = house_prices[-1]    # Last item (index -1)
print("First price:", first_price)
print("Last price:", last_price)

# --- 3. Modifying and Adding Elements ---
house_prices.append(2.8)         # Add a new value to the end
print("Updated list after adding a price:", house_prices)
```

# 3. Dictionaries & Key-Value Pairs 🗂️
**Concept:**

While a list stores items by numerical index (0, 1, 2...), a Dictionary stores data in labelled pairs called key-value pairs, written inside curly braces {}. In machine learning, a single data point (like a specific house or a user profile) is often represented as a dictionary where the keys are the feature names and the values are the measurements.





*   🏷️ Key: The name or label (usually a string, e.g., "bedrooms")

*   📌 Value: The data associated with that key (e.g., 3)



```python
# --- 1. Creating a Dictionary for a Single Sample ---
house_profile = {
    "neighborhood": "Downtown",
    "bedrooms": 3,
    "square_feet": 1450,
    "has_pool": False
}

print("Full dictionary:", house_profile)

# --- 2. Accessing Values by Key ---
# Instead of an index number, we pass the key name in brackets
rooms = house_profile["bedrooms"]
area = house_profile["square_feet"]

print("Bedrooms:", rooms)
print("Square Feet:", area)

# --- 3. Adding or Updating Keys ---
house_profile["price_estimate"] = 350000  # Adding a new key-value pair
house_profile["has_pool"] = True           # Updating an existing key
print("Updated profile:", house_profile)
```

# 4. For Loops & Iteration 🔁
**Concept:**

A For Loop automates repetitive tasks by stepping through a collection (like a list or dictionary) one item at a time. In ML workflows, loops are used to process batches of data, calculate running errors, or transform raw numbers into normalized values.


*  🚶 Iterator Variable: A temporary variable (often named item, num, or x) that holds the current element during that pass of the loop.

*   🛑 Indentation: Python uses indentation (4 spaces or 1 tab) to know which lines belong inside the loop.






```python
# --- 1. Iterating Over a List ---
# List of predicted home values in thousands
predictions = [250, 310, 420, 180]

print("Displaying predictions with currency formatting:")
for value in predictions:
    # Each loop cycle assigns the next list element to 'value'
    formatted_value = f"${value},000"
    print(formatted_value)

# --- 2. Transforming Data with a Loop ---
# Applying a 10% market adjustment to each price
adjusted_prices = []
for value in predictions:
    new_price = value * 1.10
    adjusted_prices.append(new_price)

print("Adjusted prices list:", adjusted_prices)
```


# 5. Functions ⚙️
**Concept:**

A Function is a reusable block of code that takes an input (parameters), performs a specific operation, and returns an output (return). In ML, we wrap algorithms, metric calculations, and data cleaners into functions so we can run them on any dataset without rewriting code.



*   🛠️ def keyword: Tells Python you are defining a new function.

*  📥 Parameters: The inputs placed inside parentheses.

*   📤 return statement: The result sent back to where the function was called.







```python
# --- 1. Defining a Metric Function ---
# A simple function to calculate absolute error between prediction and actual target
def calculate_absolute_error(actual, predicted):
    """Calculates the absolute difference between true and predicted values."""
    error = abs(actual - predicted)
    return error

# --- 2. Calling the Function ---
true_house_price = 300
predicted_house_price = 285

error_result = calculate_absolute_error(true_house_price, predicted_house_price)
print("Absolute error:", error_result)

# --- 3. Reusing the Function on New Values ---
print("Error for second house:", calculate_absolute_error(450, 470))
```
